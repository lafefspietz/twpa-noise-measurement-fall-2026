```python
import json
from urllib.request import urlopen
import json
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.animation as animation
```

```python
f = open('1790628601-vna-power-sweep.json')
rawdata = f.read()
f.close()
qubit_power_scan = json.loads(rawdata)
```

```python
f = open('1790630661-vna-power-sweep.json')
rawdata = f.read()
f.close()
no_qubit_power_scan = json.loads(rawdata)
```

```python
s_real_qubit = np.array(qubit_power_scan['s_real'])
s_imag_qubit = np.array(qubit_power_scan['s_imag'])
s_complex_qubit = s_real_qubit + 1j * s_imag_qubit

avg_counts = [1, 3, 10, 30, 100]
s_qubit_averaged_results = {}
s_qubit_db_results = {}
for N in avg_counts:
    s_qubit_averaged_results[N] = np.abs(np.mean(s_complex_qubit[:, :N, :], axis=1))
    s_qubit_db_results[N] = 20 * np.log10(np.abs(s_qubit_averaged_results[N]))

s_real_no_qubit = np.array(no_qubit_power_scan['s_real'])
s_imag_no_qubit = np.array(no_qubit_power_scan['s_imag'])
s_complex_no_qubit = s_real_no_qubit + 1j * s_imag_no_qubit

s_no_qubit_averaged_results = {}
s_no_qubit_db_results = {}
for N in avg_counts:
    s_no_qubit_averaged_results[N] = np.abs(np.mean(s_complex_no_qubit[:, :N, :], axis=1))
    s_no_qubit_db_results[N] = 20 * np.log10(np.abs(s_no_qubit_averaged_results[N]))

```

```python
legend = []
fig, ax = plt.subplots(figsize=(6, 6))    
plt.gca().set_prop_cycle(color=plt.cm.turbo(np.linspace(0, 1, 11)))

for index in np.arange(len(s_no_qubit_averaged_results[100])):
    if index%2 == 0:
        plt.plot(qubit_power_scan['fghz'],s_qubit_averaged_results[100][index]/s_no_qubit_averaged_results[100][index])
        legend.append("P = " + str(qubit_power_scan['power'][index]) + " dBm")


plt.xlim(qubit_power_scan['fghz'][0],qubit_power_scan['fghz'][-1])
plt.ylim(0,1.2)
plt.grid()
plt.ylabel('Normalized Transmission [linear units]')
plt.xlabel("Frequency [GHz]")
plt.title('|S_qubit|/|S_no_qubit| Power Sweep')
plt.legend(legend)
```

```python
legend = []
fig, ax = plt.subplots(figsize=(6, 6))    
plt.gca().set_prop_cycle(color=plt.cm.turbo(np.linspace(0, 1, 11)))

for index in np.arange(len(s_qubit_averaged_results[100])):
    if index%2 == 0:
        plt.plot(qubit_power_scan['fghz'],s_no_qubit_averaged_results[100][index])
        legend.append("P = " + str(qubit_power_scan['power'][index]) + " dBm")

plt.xlim(qubit_power_scan['fghz'][0],qubit_power_scan['fghz'][-1])
plt.grid()
plt.ylabel('Transmission [linear units]')
plt.xlabel("Frequency [GHz]")
plt.legend(legend)
plt.title('Qubit Biased out of Band')
plt.ylim(0,0.005)
```
