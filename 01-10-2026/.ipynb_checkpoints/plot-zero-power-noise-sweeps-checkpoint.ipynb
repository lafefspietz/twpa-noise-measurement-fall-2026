```python
import json
from urllib.request import urlopen
import json
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
files = os.listdir('.')
sweeps = []
for file in files:
    if "-noise-sweep-zero-power.json" in file:
        with open(file, 'r') as f:
            scan = json.load(f)
        sweeps.append(scan)
```

```python
len(sweeps)
```

```python


plt.figure(figsize=(8,8))
plt.tight_layout()
plt.plot(sweeps[0]['noise_bias'],np.array(sweeps[0]['noise'])/1e-9,"+")

plt.xlabel('Nominal Agilent 33220A Voltage [V]')
plt.ylabel('P [nW]')
plt.title('Just Noise, RBW = ' + str(sweeps[0]['rbw']/1e6) + ' MHz, Ref Level = ' +  f"{sweeps[0]['ref_level']/1e-9:.3f}" + " nW")
plt.grid()
```

```python
plt.figure(figsize=(8,8))
plt.gca().set_prop_cycle(color=plt.cm.turbo(np.linspace(0, 1, len(sweeps))))

plt.tight_layout()
legend = []
for index in np.arange(len(sweeps)):
    plt.plot(sweeps[index]['noise_bias'],np.array(sweeps[index]['noise'])/1e-9,"+-")
    legend.append('RBW = ' + str(sweeps[index]['rbw']/1e6) + ' MHz, Ref Level = ' +  f"{sweeps[index]['ref_level']/1e-9:.3f}" + ' nW')
plt.xlabel('Nominal Agilent 33220A Voltage [V]')
plt.ylabel('P [nW]')
plt.legend(legend)
plt.grid()
```
